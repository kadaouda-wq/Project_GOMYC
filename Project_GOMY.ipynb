```python
import tkinter as tk
import random

window = tk.Tk()
window.title("Rock Paper Scissors")
window.geometry("1200x900")

# Choix du joueur
player_choice = tk.StringVar(value="Rock")

tk.Label(window, text="Choose your move:", font=("Arial", 14)).pack(pady=10)
tk.Radiobutton(window, text="Rock", value="Rock", variable=player_choice).pack()
tk.Radiobutton(window, text="Paper", value="Paper", variable=player_choice).pack()
tk.Radiobutton(window, text="Scissors", value="Scissors", variable=player_choice).pack()

# Choix de l'ordinateur
def get_computer_choice():
    return random.choice(["Rock", "Paper", "Scissors"])

# Gagnant
def determine_winner(player, computer):
    if player == computer:
        return "Draw"
    if (player == "Rock" and computer == "Scissors") or \
       (player == "Scissors" and computer == "Paper") or \
       (player == "Paper" and computer == "Rock"):
        return "You win!"
    else:
        return "Computer wins!"

# Résultats
result_label = tk.Label(window, text="", font=("Arial", 14))
result_label.pack(pady=10)

# Scoreboard
scores = {"Player": 0, "Computer": 0, "Draws": 0}
score_label = tk.Label(window, text="Player: 0 | Computer: 0 | Draws: 0", font=("Arial", 12))
score_label.pack()

def update_score(result):
    if result == "You win!":
        scores["Player"] += 1
    elif result == "Computer wins!":
        scores["Computer"] += 1
    else:
        scores["Draws"] += 1

    score_label.config(
        text=f"Player: {scores['Player']} | Computer: {scores['Computer']} | Draws: {scores['Draws']}"
    )

def play_game():
    player = player_choice.get()
    computer = get_computer_choice()
    result = determine_winner(player, computer)

    update_score(result)

    result_label.config(
        text=f"You chose: {player}\nComputer chose: {computer}\nResult: {result}"
    )

# Bouton Play
tk.Button(window, text="Play", font=("Arial", 12), command=play_game).pack(pady=15)

window.mainloop()
```
